# Goal
- Multivariate Linear regression (implementation only)  without Library

```python
import numpy as np
import matplotlib.pyplot as plot
np.set_printoptions(precision=2)  # reduced display precision on numpy arrays
```

```python
# Prediction for a single example
def predict(x, w, b):
    
    """
    single predict using linear regression
    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters   
      b (scalar):             model parameter 
      
    Returns:
      p (scalar):  prediction
    """

    return np.dot(x, w) + b

# Prediction for entire training set
def predict_all(X, w, b):
    f_wb = X @ w + b # or can write np.dot(X, w) + b
    return f_wb
```

```python
# Computing cost for training set at some w & b
def compute_cost(X, y, w, b):
    
    """
    compute cost
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters  
      b (scalar)       : model parameter
      
    Returns:
      cost (scalar): cost
    """

    m = X.shape[0]
    f_wb = predict_all(X, w, b)
    cost = np.sum((f_wb - y) ** 2) / (2 * m)
    return cost
```

```python
# Gradient vector -> dj_dw & Gradient -> dj_db at some w & b

def compute_gradient(X, y, w, b):
    m= X.shape[0]
    f_wb = predict_all(X, w, b)
    error = f_wb - y

    # Removed all the loop, and doing heavy use of vectorization
    dj_db = np.sum(error) / m
    dj_dw = np.dot(X.T, error) / m # or equivalently (X.T @ error) / m

    return dj_dw, dj_db
```

```python
# Z-Score for normalization
# axis = 0 means column wise, axis = 1 means row wise

def zscore_normalize_features(X_train):
    """
    computes  X, z-score normalized by column
    
    Args:
      X_train (ndarray (m,n))     : input data, m examples, n features
      
    Returns:
      X_norm (ndarray (m,n)): input normalized by column
      mean (ndarray (n,))     : mean of each feature
      sigma (ndarray (n,))  : standard deviation of each feature
    """
    
    mean = np.mean(X_train, axis=0)
    sigma = np.std(X_train, axis=0)
    sigma = np.where(sigma == 0, 1, sigma)
    X_norm = (X_train - mean ) / sigma # Broadcasting of numpy arrays

    return X_norm, mean, sigma

def normalize(x, mean, sigma):
    return (x - mean) / sigma
```

```python
# Implementation of GD for multiple features

def gradient_descent(X, y, w_init, b_init, alpha, iterations, stopping_criteria = 1e-6):
    
    w, b = w_init.copy(), b_init

    for i in range(iterations):            
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        w -= alpha * dj_dw
        b -= alpha * dj_db
        if compute_cost(X, y, w, b) < stopping_criteria: # Stop when cost becomes sufficiently small
            break
        
    return w, b
```

```python
X_train = np.array([ # Large feature, needs rescaling
    [2104, 5, 1, 45],
    [1416, 3, 2, 40],
    [852, 2, 1, 35]
], dtype=float)
y_train = np.array([460, 232, 178])

# X_train = np.array([  # [size, bedrooms, bathrooms, age]
#     [2, 1, 1, 5],
#     [3, 2, 1, 4],
#     [4, 2, 2, 3],
#     [5, 3, 2, 2],
#     [6, 3, 3, 2],
#     [7, 4, 3, 1]
# ], dtype=float) 
# y_train = np.array([ 15, 19, 26, 31, 37, 41 ], dtype=float) # House prices
```

```python
# ===== Choose this if needs Normalization =====

alpha, iterations = 0.1, 1000

# Parameters initialization, can choose a better starting point for faster convergence
w_init = np.array([0., 0., 0., 0.])
b_init = 0.
stopping_criteria = 1e-6
    
# Normalization step =========
X_norm, mean, sigma = zscore_normalize_features(X_train)

# Calling GD ==========
w_final, b_final = gradient_descent(X_norm, y_train, w_init, b_init, alpha, iterations, stopping_criteria)

# Details of convergence ==========
print("===== After convergence =====")
print(f"\tw = {w_final}")
print(f"\tb = {b_final}")
print(f"\tCost after convergence = {compute_cost(X_norm, y_train, w_final, b_final)}")
print(f"\ty_train = {y_train}")
print(f"\tf_wb = {predict_all(X_norm, w_final, b_final)}\n")

# Testing on a dataset ===========
print("===== Performing Predictions =====")
x_test = np.array([1416, 3, 2, 40], dtype=float)
x_norm = normalize(x_test, mean, sigma) # Mandatory to first scale down the testing dataset
print(f"\tPrediction for x_test is {predict(x_norm, w_final, b_final)}")
```
