```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([2, 6]) #in 1000's
y_train = np.array([6, 18]) #in crores
m = x_train.shape[0] # or .len()
```

```python
plt.scatter(x_train, y_train, marker= 'x', c='b', label='Original y') # c-> color
plt.xlabel("Quantity in 1000's")
plt.ylabel("Amount in crores")
plt.title('Laptop Pricing')
plt.legend() # Shows label inside the plot
plt.show()
```

```python
def compute_model_output(x_train, w, b): # For computing Predicted y for w & b
    m = len(x_train)
    f_wb = np.zeros(m)
    for i in range(m):
        x_i = x_train[i]
        f_wb[i] = w*x_i + b
    return f_wb
```

```python
f_wb = compute_model_output(x_train, w=3, b=10) # Computing predicted y
plt.scatter(x_train, y_train, marker='x', c='b', label='Original parameters')
plt.scatter(x_train, f_wb, marker='o', c='r', label='Predicted parameters')
plt.plot(x_train, f_wb, c='g') # plot joins consecutive points with a line segment
plt.xlabel("Quantity in 1000's")
plt.ylabel('Price in crores')
plt.title('Laptop Pricing')
plt.legend()
plt.show()
```
