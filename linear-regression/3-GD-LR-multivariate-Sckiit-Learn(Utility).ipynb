# Goal
- Multivariate Linear Regression (implementation only) Using Scikit learn library
- It uses Closed form solution for this

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
```

```python
X_train = np.array([ # Large feature, needs rescaling
    [2104, 5, 1, 45],
    [1416, 3, 2, 40],
    [852, 2, 1, 35]
], dtype=float)
y_train = np.array([460, 232, 178])

# X_train = np.array([  # [size, bedrooms, bathrooms, age]
#     [2, 1, 1, 5],
#     [3, 2, 1, 4],
#     [4, 2, 2, 3],
#     [5, 3, 2, 2],
#     [6, 3, 3, 2],
#     [7, 4, 3, 1]
# ], dtype=float) 
# y_train = np.array([ 15, 19, 26, 31, 37, 41 ], dtype=float) # House prices
```

```python
# Normalization

scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
mean, sigma = scaler.mean_, scaler.scale_

print("Without Normalization -> X_train ")
print(X_train)
print()
    
print("After Normalization -> X_train ")
print(X_norm)
print(f"\nMean: {mean}")
print(f"Standard deviation: {sigma}")
print()
```

```python
# Gradient descent

sgdr = SGDRegressor(max_iter=100000)
sgdr.fit(X_norm, y_train)
w = sgdr.coef_
b = sgdr.intercept_
print(f"Coefficients of linear reg (w) -> ", w)
print(f"Intercept (b) -> ", b)
```

```python
# Make predictions on training set
y_pred = sgdr.predict(X_norm)
print("Prediction on X_train ===")
print(f"\tActual y -> {y_train}")
print(f"\tPredicted y -> {y_pred}")
print(f"\tw.x + b -> {np.dot(X_norm, w) + b}\n")

# Prediction on a sample
x_test, y_test= X_train[0].reshape(1, -1), y_train[0] # Transforming vector to matrix
prediction = sgdr.predict(scaler.transform(x_test))[0] # Prediction on a matrix, after normalization
print("Prediction on x_test ===")
print(f"\tTest set {x_test[0]}")
print(f"\tActual y: {y_test}")
print(f"\tPredicted y: {prediction}")
```
