```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Training data
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])
```

```python

# Create and train the Logistic Regression model
lr_model = LogisticRegression(max_iter=10000)
lr_model.fit(X_train, y_train)

# Prediction on training data
y_pred = lr_model.predict(X_train)
print(f"Predictions on training data: {y_pred}")

# Prediction on new/unseen examples
X_new = [[1.5, 2.], [2., 0.7]]
y_pred_new = lr_model.predict(X_new)
print(f"Predictions on new examples: {y_pred_new}")

# Model parameters
print(f"weights: {lr_model.coef_}")
print(f"Bias: {lr_model.intercept_}")

# Calculating accuracy
accuracy = lr_model.score(X_train, y_train)
print(f"Training Accuracy: {accuracy * 100:.2f}%")
```
